# Recherche d'information sur le corpus CISI — Système 1 : TF-IDF

**Auteurs :** Melissa Boko et Amadou Bah — projet de Traitement Automatique des Langues, INSA Rennes

**Objectif :** pour chaque requête du corpus CISI, retrouver les documents pertinents parmi 1 460 résumés d'articles scientifiques.

**Approche :** modèle vectoriel classique.
1. Prétraitement linguistique avec spaCy : lemmatisation, suppression des mots vides, conservation des noms, verbes et adjectifs uniquement.
2. Filtrage du vocabulaire selon la fréquence documentaire (termes trop rares ou trop fréquents écartés).
3. Pondération **TF-IDF** et construction d'un **index inversé**.
4. Classement des documents par **similarité cosinus** avec la requête, avec un seuil de pertinence.

Ce système sert de référence pour le système 2 (recherche dense et re-ranking neuronal).

```python
import os
import re
import math
from collections import Counter, defaultdict

import spacy
```

```python
# Fichiers d'entrée (à placer dans /content sous Colab)
CORPUS_PATH = "/content/CISI.ALLnettoye"
QRY_PATH = "/content/CISI.QRY"
REL_PATH = "/content/CISI_dev.REL"
EVAL_SCRIPT_PATH = "/content/eval.pl"

# Fichier de résultats produit par ce système
RUN_PATH = "/content/run_tfidf.REL"

# Paramètres du vocabulaire
DF_MIN = 2            # un terme doit apparaître dans au moins 2 documents
DF_MAX_RATIO = 0.80   # et dans au plus 80 % des documents

# Paramètres de la recherche
TOP_K = 50            # nombre maximum de documents renvoyés par requête
SEUIL_SCORE = 0.12    # similarité cosinus minimale pour qu'un document soit renvoyé

nlp = spacy.load("en_core_web_md")
print("Pipeline spaCy :", nlp.pipe_names)
for chemin in [CORPUS_PATH, QRY_PATH, REL_PATH, EVAL_SCRIPT_PATH]:
    print(chemin, "->", "trouvé" if os.path.exists(chemin) else "MANQUANT")
```

## 1. Parsing du corpus

Chaque document commence par une ligne `.I <numéro>` suivie de son texte.

```python
def parser_collection_cisi(chemin_fichier):
    documents = {}
    id_document_courant = None
    lignes_document_courant = []

    def enregistrer():
        texte = re.sub(r"\s+", " ", " ".join(lignes_document_courant)).strip()
        documents[id_document_courant] = texte

    with open(chemin_fichier, "r", encoding="utf-8", errors="ignore") as fichier:
        for ligne in fichier:
            ligne = ligne.strip()
            if ligne.startswith(".I "):
                if id_document_courant is not None:
                    enregistrer()
                id_document_courant = int(ligne.split()[1])
                lignes_document_courant = []
            else:
                lignes_document_courant.append(ligne)

    if id_document_courant is not None:
        enregistrer()

    return documents
```

## 2. Prétraitement et indexation TF-IDF

**Normalisation d'un token :** on ne garde que les mots alphabétiques, hors mots vides, appartenant aux catégories grammaticales porteuses de sens (noms, noms propres, verbes, adjectifs), ramenés à leur lemme en minuscules.

**Pondération :**
- TF logarithmique : $tf = 1 + \log(n)$, qui atténue l'effet des mots répétés ;
- IDF lissé : $idf = \log_{10}\frac{N + 1}{df + 1} + 1$.

```python
CATEGORIES_GARDEES = {"NOUN", "PROPN", "VERB", "ADJ"}


def normaliser_token(token):
    if token.is_space or token.is_punct or not token.is_alpha or token.is_stop:
        return None
    if token.pos_ not in CATEGORIES_GARDEES:
        return None

    lemme = token.lemma_.lower().strip()
    if not lemme or lemme == "-pron-":
        lemme = token.text.lower().strip()
    if len(lemme) < 2:
        return None
    return lemme


def pretraiter_textes(textes_par_id, nlp, taille_lot=64):
    """Transforme chaque texte en liste de termes normalisés (utilisé pour les documents et les requêtes)."""
    ids = list(textes_par_id.keys())
    textes = [textes_par_id[i] for i in ids]
    resultat = {}
    for id_texte, doc_spacy in zip(ids, nlp.pipe(textes, batch_size=taille_lot)):
        termes = [normaliser_token(token) for token in doc_spacy]
        resultat[id_texte] = [t for t in termes if t is not None]
    return resultat


def calculer_frequences_documentaires(documents_traites):
    frequences = Counter()
    for termes in documents_traites.values():
        frequences.update(set(termes))
    return frequences


def construire_vocabulaire_filtre(frequences_documentaires, nombre_documents, df_min, df_max_ratio):
    return {terme for terme, df in frequences_documentaires.items()
            if df_min <= df <= df_max_ratio * nombre_documents}


def construire_structures_tfidf(documents_traites, frequences_documentaires, vocabulaire):
    nombre_documents = len(documents_traites)
    idf = {terme: math.log10((nombre_documents + 1) / (frequences_documentaires[terme] + 1)) + 1
           for terme in vocabulaire}

    vecteurs_documents = {}
    index_inverse = defaultdict(list)

    for id_document, termes in documents_traites.items():
        vecteur = {}
        for terme, compte in Counter(termes).items():
            if terme not in vocabulaire:
                continue
            poids = (1 + math.log(compte)) * idf[terme]
            vecteur[terme] = poids
            index_inverse[terme].append((id_document, poids))
        vecteurs_documents[id_document] = vecteur

    return idf, vecteurs_documents, dict(index_inverse)
```

```python
documents = parser_collection_cisi(CORPUS_PATH)
documents_traites = pretraiter_textes(documents, nlp)
frequences_documentaires = calculer_frequences_documentaires(documents_traites)
vocabulaire = construire_vocabulaire_filtre(frequences_documentaires, len(documents_traites), DF_MIN, DF_MAX_RATIO)
idf, vecteurs_documents, index_inverse = construire_structures_tfidf(documents_traites, frequences_documentaires, vocabulaire)

print("Nombre de documents :", len(documents))
print("Taille du vocabulaire filtré :", len(vocabulaire))
print("Document 1 après prétraitement :", documents_traites[1][:30])
```

## 3. Requêtes et recherche par similarité cosinus

Seules les sections titre (`.T`) et texte (`.W`) des requêtes sont conservées. Les requêtes subissent exactement le même prétraitement que les documents.

L'**index inversé** permet de ne calculer le score que pour les documents qui partagent au moins un terme avec la requête, au lieu de parcourir tout le corpus.

```python
def parser_requetes_cisi(chemin_fichier, sections_a_garder=("T", "W")):
    requetes = {}
    id_requete_courante = None
    section_courante = None
    lignes = []

    def enregistrer():
        requetes[id_requete_courante] = re.sub(r"\s+", " ", " ".join(lignes)).strip()

    with open(chemin_fichier, "r", encoding="utf-8", errors="ignore") as fichier:
        for ligne in fichier:
            ligne = ligne.strip()
            if ligne.startswith(".I "):
                if id_requete_courante is not None:
                    enregistrer()
                id_requete_courante = int(ligne.split()[1])
                section_courante = None
                lignes = []
            elif ligne in {".T", ".A", ".W", ".B"}:
                section_courante = ligne[1:]
            elif section_courante in sections_a_garder:
                lignes.append(ligne)

    if id_requete_courante is not None:
        enregistrer()

    return requetes


def construire_vecteurs_requetes(requetes_traitees, vocabulaire, idf):
    vecteurs = {}
    for id_requete, termes in requetes_traitees.items():
        compte = Counter(t for t in termes if t in vocabulaire)
        vecteurs[id_requete] = {t: (1 + math.log(n)) * idf[t] for t, n in compte.items()}
    return vecteurs


def norme(vecteur):
    return math.sqrt(sum(p * p for p in vecteur.values()))


def rechercher_une_requete(vecteur_requete, index_inverse, normes_documents, top_k, seuil_score):
    norme_requete = norme(vecteur_requete)
    if norme_requete == 0:
        return []

    # Produit scalaire calculé uniquement via l'index inversé
    produits_scalaires = defaultdict(float)
    for terme, poids_requete in vecteur_requete.items():
        for id_document, poids_document in index_inverse.get(terme, []):
            produits_scalaires[id_document] += poids_requete * poids_document

    resultats = []
    for id_document, produit in produits_scalaires.items():
        norme_document = normes_documents.get(id_document, 0.0)
        if norme_document == 0:
            continue
        score = produit / (norme_requete * norme_document)
        if score >= seuil_score:
            resultats.append((id_document, score))

    resultats.sort(key=lambda x: x[1], reverse=True)
    return resultats[:top_k]


def ecrire_run(resultats, chemin_sortie):
    with open(chemin_sortie, "w", encoding="utf-8") as fichier:
        for id_requete in sorted(resultats):
            for id_document, score in resultats[id_requete]:
                fichier.write(f"{id_requete} {id_document} {score:.12f}\n")
```

## 4. Exécution

```python
requetes = parser_requetes_cisi(QRY_PATH)
requetes_traitees = pretraiter_textes(requetes, nlp)
vecteurs_requetes = construire_vecteurs_requetes(requetes_traitees, vocabulaire, idf)
normes_documents = {id_doc: norme(v) for id_doc, v in vecteurs_documents.items()}

resultats = {
    id_requete: rechercher_une_requete(vecteur, index_inverse, normes_documents, TOP_K, SEUIL_SCORE)
    for id_requete, vecteur in vecteurs_requetes.items()
}

ecrire_run(resultats, RUN_PATH)
print(f"{len(resultats)} requêtes traitées — résultats écrits dans {RUN_PATH}")
print(f"Configuration : TOP_K = {TOP_K}, seuil = {SEUIL_SCORE}")
```
